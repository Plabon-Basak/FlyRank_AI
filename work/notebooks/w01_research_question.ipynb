## 2. The question: decision, action, cost of a wrong call

*What decision does your work improve? Who acts on it? What does a wrong recommendation cost?*

The decision is **which pages should be reviewed first**. A content or SEO reviewer would use the ranked output to decide where to spend limited review time, such as checking whether a page needs a refresh, CTR review, engagement review, or monitoring. A wrong recommendation has two main costs: a reviewer may spend time on a low-priority page, or an important page may be missed and reviewed later. Therefore, the output should support human prioritization rather than automatically publish or change content.


```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
required_columns = [
    "content_id",
    "impressions_90d",
    "sessions_90d",
    "content_age_days",
    "trend_direction"
]

missing = [c for c in required_columns if c not in df.columns]

print("Required columns present:", len(missing) == 0)

if missing:
    print("Missing columns:", missing)
else:
    print("Decision grain: one content page")
```

## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

The starter data contains **30,000 page rows**. Of these, **16,262 pages are labeled as declining**, which is a declining-label rate of about **54.2%**. This gives enough variation in the starter data to investigate whether observable page and search signals can help prioritize pages for review. These numbers are evidence that the problem is worth investigating, not proof that a future model will work on the full warehouse.


```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
df["is_declining_label"] = (
    df["trend_direction"]
    .astype(str)
    .str.lower()
    .eq("down")
    .astype(int)
)

total_pages = len(df)
declining_pages = int(df["is_declining_label"].sum())
declining_rate = df["is_declining_label"].mean()

print(f"Total starter pages: {total_pages:,}")
print(f"Pages with down trend: {declining_pages:,}")
print(f"Declining-label rate: {declining_rate:.3f}")
```

## 4. Careful words: what I can and can't claim

*Write what your work will be able to say (observed, directional, decision-support) — and what it never will (causal proof, 'predicting Google').*

This project can make **observed and measured** statements about the signals in the available data and can provide **directional, decision-support** recommendations for which pages deserve human review first. The starter declining label is based on the observed `trend_direction` field, so it should be treated as a proxy rather than a perfect future outcome. I cannot claim that the model proves that a content change will cause recovery, that a page is guaranteed to improve after a refresh, or that the model is predicting Google's ranking decisions. A stronger later version should use clearly separated historical feature windows and future outcome windows to reduce leakage and make the evaluation more meaningful.
